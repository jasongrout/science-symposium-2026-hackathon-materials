# 01 — Exposure time calculator

`slicersim` simulates what the **Lazuli** space telescope's integral field spectrograph
(IFS) would actually observe: a realistic spectrum, with realistic noise, for any target
you give it. The question every observer asks first is *"how long do I need to expose to
get the data quality I need?"* — this notebook answers exactly that, in two or three
lines of code, for both a supernova and a star.

This is notebook **1 of 5** in the beginner tutorial series:

1. **Exposure time calculator** (this notebook)
2. Getting a spectrum or a cube, and reading the instrument configuration
3. Changing target and detector properties
4. Where the noise comes from (the variance budget)
5. The Lazuli instrument: two channels, two cubes

Let's start.

## 1. The fastest path: a two-line supernova ETC

`slicersim.lazuli_sn_etc(...)` builds a SN Ia at a given redshift, configures the
detector so that the target signal-to-noise ratio is reached over a chosen wavelength
window, and returns the exposure time together with the configured target object.

```python
import numpy as np
import matplotlib.pyplot as plt
import slicersim
```

```python
%%time
exptime, sn_target = slicersim.lazuli_sn_etc(
    "salt", redshift=0.5, x1=0.0, c=0.0,
    snr=15, per_resolution=True,
    lbda_range=[4000, 7000], frame="rest",
)
print(f"exposure time: {exptime:.1f} s ({exptime/3600:.2f} h)")
```

`sn_target` is a ready-to-use `LazuliSupernova` object, already configured to reach
`snr=15` averaged (per resolution element) between 4000 A and 7000 A rest-frame.

## 2. The same idea for a star

`slicersim.LazuliCalSpec` loads a flux-calibrated CalSpec standard star spectrum
directly by name. `setup_to_snr()` is the underlying method that both `lazuli_sn_etc`
and `lazuli_etc` call for you — here we call it ourselves, which is just as short.

```python
star = slicersim.LazuliCalSpec("bd_17")

# reach snr=30 between 4000A and 8000A (observer frame, the default)
readout_config, reached_snr = star.setup_to_snr(30, lbda_range=[4_000, 8_000])
print(f"reached snr={reached_snr:.1f} with readout config: {readout_config}")
```

`setup_to_snr()` searches over the detector's read-out modes (how many groups,
frames per group, and ramps to use) and returns the configuration it converged on,
together with the SNR actually reached — it will not be exactly your target value,
but close.

## 3. `get_exposure_time()`

Once a target is configured (by `setup_to_snr`, or manually — see notebook 3), the
resulting exposure time is always available with `get_exposure_time()`.

```python
print(f"star exposure time: {star.get_exposure_time():.1f} s")
print(f"SN exposure time:   {sn_target.get_exposure_time():.1f} s")
```

Pass `full=True` to see the full timing breakdown: frame time, group time, and the
total exposure time (integration time plus overheads).

```python
sn_target.get_exposure_time(full=True)
```

## 4. `get_data_volume()`

Longer exposures with more read-out groups also mean more raw data. `get_data_volume()`
returns the size of the corresponding data (all the individual detector reads, not just
the final combined spectrum).

```python
for name, target in [("star", star), ("SN Ia", sn_target)]:
    print(f"{name:6s}: {target.get_exposure_time()/3600:5.2f} h  ->  {target.get_data_volume(units='GB'):.2f} GB")
```

## 5. Any spectrum, any magnitude

You are not limited to the built-in `LazuliSupernova`/`LazuliCalSpec` convenience
classes. `slicersim.lazuli_etc()` takes *any* input spectrum (wavelength + flux arrays)
and a target magnitude in any band, and returns the exposure time exactly the same way.

```python
# a flat spectrum, normalized to mag=21 in sdss-r
lbda = np.arange(3_000, 20_000, 0.5)  # every 0.5A, oversampled
flux = np.ones_like(lbda)

exptime_flat, flat_target = slicersim.lazuli_etc(
    lbda, flux, snr=20, per_resolution=True,
    lbda_range=[7_000, 8_000], mag=21, band="sdssr",
)
print(f"exposure time for a flat, mag=21 (sdssr) source: {exptime_flat:.1f} s")
```

```python
fig, ax = plt.subplots(figsize=(5, 3.5))
names = ["star\n(BD+17, snr=30)", "SN Ia\n(z=0.5, snr=15)", "flat spectrum\n(mag=21, snr=20)"]
times_h = [star.get_exposure_time()/3600, sn_target.get_exposure_time()/3600, exptime_flat/3600]
ax.bar(names, times_h, color=["#194D80", "#A8B7C7", "#E07A5F"])
ax.set_ylabel("exposure time [hours]")
fig.tight_layout()
```

## Recap

- `slicersim.lazuli_sn_etc(...)` and `slicersim.lazuli_etc(...)` are the fastest paths:
  one function call gives you both the exposure time and a ready-to-use target.
- `LazuliSupernova`, `LazuliCalSpec` (and the generic `LazuliTarget`) are the underlying
  objects — call `target.setup_to_snr(snr, lbda_range=..., frame=...)` yourself for more
  control.
- `target.get_exposure_time()` and `target.get_data_volume()` give you the cost, in time
  and in bytes, of any configured target.
